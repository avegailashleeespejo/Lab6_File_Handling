```python

# EXERCISE 3: CYBERSECURITY INCIDENT REPORT

import csv
import json
import shutil
from pathlib import Path

# Student-specific parameters
SEED_DIGIT = 9
ID_SUM = 21

# Target year calculation
TARGET_YEAR = 2015 + SEED_DIGIT

# Dataset location
INPUT_FILE = Path(
    "Global_Cybersecurity_Threats_2015-2024.csv"
)

# Output directory
OUTPUT_FOLDER = Path("output")
OUTPUT_FOLDER.mkdir(exist_ok=True)

print("CYBERSECURITY INCIDENT REPORT")

print("SEED_DIGIT:", SEED_DIGIT)
print("ID_SUM:", ID_SUM)
print("Target Year:", TARGET_YEAR)

```

```python

# CELL 2: READ AND INSPECT DATASET

if not INPUT_FILE.exists():

    print("ERROR: Dataset file not found!")
    print("Please place the CSV file inside")
    print("the Exercise 3 folder.")

else:

    with open(
        INPUT_FILE,
        "r",
        newline="",
        encoding="utf-8-sig"
    ) as file:

        reader = csv.DictReader(file)

        print("CSV file found successfully!")

        print("\nCOLUMN NAMES:")
        print(reader.fieldnames)

        print("\nFIRST 3 RECORDS:")

        for index, row in enumerate(reader):

            if index == 3:
                break

            print(row)
```

```python

# CELL 3: FILTER AND ANALYZE CYBERSECURITY INCIDENTS

from collections import Counter
import csv

# Dataset column names
YEAR_COL = "Year"
ATTACK_COL = "Attack Type"
LOSS_COL = "Financial Loss (in Million $)"

# Initialize counters
total_records = 0
invalid_year_records = 0
invalid_loss_records = 0

target_incidents = []
attack_counts = Counter()

total_financial_loss = 0.0
valid_loss_count = 0

# Check dataset
if not INPUT_FILE.exists():
    raise FileNotFoundError(
        f"Dataset not found: {INPUT_FILE.resolve()}"
    )

# Read dataset line by line
with open(
    INPUT_FILE,
    "r",
    newline="",
    encoding="utf-8-sig"
) as file:

    reader = csv.DictReader(file)

    required_columns = [
        YEAR_COL,
        ATTACK_COL,
        LOSS_COL
    ]

    if not reader.fieldnames:
        raise ValueError("Dataset has no column headers.")

    missing_columns = [
        col for col in required_columns
        if col not in reader.fieldnames
    ]

    if missing_columns:
        raise ValueError(
            f"Missing columns: {missing_columns}"
        )

    original_fields = reader.fieldnames

    # Process each record
    for index, row in enumerate(reader):

        total_records += 1

        # Check year
        try:
            incident_year = int(row[YEAR_COL])

        except (ValueError, TypeError):
            invalid_year_records += 1
            continue

        # Select incidents in target year
        if incident_year != TARGET_YEAR:
            continue

        target_incidents.append(row)

        # Count attack types
        attack_type = (
            row.get(ATTACK_COL) or "Unknown"
        ).strip()

        attack_counts[attack_type] += 1

        # Calculate financial loss
        try:
            loss = float(row[LOSS_COL])

            if loss >= 0:
                total_financial_loss += loss
                valid_loss_count += 1

        except (ValueError, TypeError):
            invalid_loss_records += 1

# Calculate total and average financial loss
incident_count = len(target_incidents)

average_financial_loss = (
    total_financial_loss / valid_loss_count
    if valid_loss_count > 0
    else 0
)

# Prepare valid incidents for ranking
ranked_incidents = []

for index, row in enumerate(target_incidents):

    try:
        loss = float(row[LOSS_COL])

        if loss >= 0:
            ranked_incidents.append((index, row, loss))

    except (ValueError, TypeError):
        continue

# Sort by descending financial loss.
# Python's stable sort retains original order when tied.
ranked_incidents.sort(
    key=lambda item: item[2],
    reverse=True
)

# Select five costliest incidents
top_five_incidents = [
    row for index, row, loss in ranked_incidents[:5]
]

# Export all target-year incidents to CSV
filtered_csv = OUTPUT_FOLDER / "filtered_incidents.csv"

with open(
    filtered_csv,
    "w",
    newline="",
    encoding="utf-8"
) as file:

    writer = csv.DictWriter(
        file,
        fieldnames=original_fields
    )

    writer.writeheader()
    writer.writerows(target_incidents)

# Display results
print(" CYBERSECURITY INCIDENT ANALYSIS")

print("\nSTUDENT PARAMETERS")
print("SEED_DIGIT:", SEED_DIGIT)
print("ID_SUM:", ID_SUM)
print("Target Year:", TARGET_YEAR)

print("\nDATASET INFORMATION")
print("Total records:", total_records)
print("Invalid year records:", invalid_year_records)
print("Incidents in target year:", incident_count)

print("\nATTACK TYPE COUNTS")

for attack, count in sorted(attack_counts.items()):
    print(attack, ":", count)

print("\nFINANCIAL LOSS RESULTS")
print(
    "Total financial loss:",
    round(total_financial_loss, 2),
    "Million USD"
)

print(
    "Average financial loss:",
    round(average_financial_loss, 2),
    "Million USD"
)

print("Invalid financial values:", invalid_loss_records)

print("\nFIVE COSTLIEST INCIDENTS")

for number, row in enumerate(top_five_incidents, start=1):

    print(
        number,
        row[ATTACK_COL],
        "-",
        row[LOSS_COL],
        "Million USD"
    )

print("\nFILTERED CSV")
print("File:", filtered_csv)
print("Records exported:", incident_count)

```

```python

# CELL 4: JSON SUMMARY AND BACKUP

import json
import shutil

# Create JSON summary
summary = {
    "student_information": {
        "seed_digit": SEED_DIGIT,
        "id_sum": ID_SUM
    },

    "target_year": TARGET_YEAR,

    "dataset_information": {
        "total_records": total_records,
        "incidents_in_target_year": incident_count,
        "invalid_year_records": invalid_year_records,
        "invalid_financial_values": invalid_loss_records
    },

    "attack_type_counts": dict(attack_counts),

    "financial_summary": {
        "total_financial_loss_million_usd": round(
            total_financial_loss, 2
        ),
        "average_financial_loss_million_usd": round(
            average_financial_loss, 2
        ),
        "valid_financial_records": valid_loss_count
    },

    "five_costliest_incidents": top_five_incidents
}

# Save JSON file
json_file = OUTPUT_FOLDER / "cybersecurity_summary.json"

with open(
    json_file,
    "w",
    encoding="utf-8"
) as file:

    json.dump(summary, file, indent=4)

print("JSON SUMMARY CREATED")
print("File:", json_file)

# Create backup of JSON file
backup_file = OUTPUT_FOLDER / "cybersecurity_summary_backup.json"

shutil.copy2(json_file, backup_file)

# Verify JSON backup
print("\nBACKUP VERIFICATION")

if backup_file.exists():

    with open(json_file, "rb") as original:
        original_content = original.read()

    with open(backup_file, "rb") as backup:
        backup_content = backup.read()

    if original_content == backup_content:
        print("Backup verification: SUCCESS")
    else:
        print("Backup verification: FAILED")

    print("Original JSON size:", json_file.stat().st_size, "bytes")
    print("Backup JSON size:", backup_file.stat().st_size, "bytes")

else:
    print("Backup file was not created.")

# Verify JSON can be read
with open(json_file, "r", encoding="utf-8") as file:
    verified_summary = json.load(file)

print("\nJSON READ VERIFICATION")
print("JSON file is valid:", isinstance(verified_summary, dict))
print(
    "Five costliest incidents:",
    len(verified_summary["five_costliest_incidents"])
)
```

```python

# CELL 5: GENERATE FINAL TEXT REPORT

report_file = OUTPUT_FOLDER / "cybersecurity_report.txt"

report_lines = []

report_lines.append("       CYBERSECURITY INCIDENT REPORT")

report_lines.append("")
report_lines.append("STUDENT INFORMATION")
report_lines.append(f"SEED_DIGIT: {SEED_DIGIT}")
report_lines.append(f"ID_SUM: {ID_SUM}")
report_lines.append(f"Target Year: {TARGET_YEAR}")

report_lines.append("")
report_lines.append("DATASET INFORMATION")
report_lines.append(f"Total records: {total_records}")
report_lines.append(f"Incidents in target year: {incident_count}")
report_lines.append(f"Invalid year records: {invalid_year_records}")

report_lines.append("")
report_lines.append("ATTACK TYPE COUNTS")

for attack, count in sorted(attack_counts.items()):
    report_lines.append(f"{attack}: {count}")

report_lines.append("")
report_lines.append("FINANCIAL LOSS SUMMARY")

report_lines.append(
    f"Total financial loss: {total_financial_loss:.2f} Million USD"
)

report_lines.append(
    f"Average financial loss: {average_financial_loss:.2f} Million USD"
)

report_lines.append(
    f"Invalid financial values: {invalid_loss_records}"
)

report_lines.append("")
report_lines.append("FIVE COSTLIEST INCIDENTS")

for number, row in enumerate(top_five_incidents, start=1):

    report_lines.append(
        f"{number}. {row[ATTACK_COL]} - "
        f"{row[LOSS_COL]} Million USD"
    )

    report_lines.append(
        f"   Country: {row.get('Country', 'N/A')}"
    )

report_lines.append("")
report_lines.append("OUTPUT FILE INFORMATION")
report_lines.append(f"Filtered CSV: {filtered_csv}")
report_lines.append(f"JSON Summary: {json_file}")
report_lines.append(f"JSON Backup: {backup_file}")

report_lines.append("")
report_lines.append("FILE VERIFICATION")
report_lines.append(f"CSV exists: {filtered_csv.exists()}")
report_lines.append(f"JSON exists: {json_file.exists()}")
report_lines.append(f"Backup exists: {backup_file.exists()}")

# Save text report
with open(
    report_file,
    "w",
    encoding="utf-8"
) as file:

    file.write("\n".join(report_lines))

# Display final report
print("\n".join(report_lines))

```
