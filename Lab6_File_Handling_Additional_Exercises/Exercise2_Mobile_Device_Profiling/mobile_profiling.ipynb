```python

# EXERCISE 2: MOBILE DEVICE USAGE PROFILING

import csv
import json
import shutil
import os
from pathlib import Path

# Student-specific parameters
SEED_DIGIT = 9
ID_SUM = 21

# Generated requirements
TARGET_OS = "iOS"
TARGET_BEHAVIOR_CLASS = (SEED_DIGIT % 5) + 1

# File locations
INPUT_FILE = Path("user_behavior_dataset.csv")
OUTPUT_FOLDER = Path("output")

# Create output directory
OUTPUT_FOLDER.mkdir(exist_ok=True)

print("MOBILE DEVICE USAGE PROFILING")
print("SEED_DIGIT:", SEED_DIGIT)
print("ID_SUM:", ID_SUM)
print("Assigned Operating System:", TARGET_OS)
print("Target Behavior Class:", TARGET_BEHAVIOR_CLASS)
print("Output folder created successfully!")
```

```python

# CELL 2: READ AND INSPECT MOBILE DATASET

if not INPUT_FILE.exists():

    print("ERROR: Dataset file not found!")
    print("Please place user_behavior_dataset.csv")
    print("inside the Exercise 2 folder.")

else:

    with open(
        INPUT_FILE,
        "r",
        newline="",
        encoding="utf-8-sig"
    ) as file:

        reader = csv.DictReader(file)

        print("CSV file found successfully!")

        print("\nCOLUMN NAMES:")
        print(reader.fieldnames)

        print("\nFIRST 3 RECORDS:")

        for index, row in enumerate(reader):

            if index == 3:
                break

            print(row)
```

```python

# CELL 3: FILTER USERS AND CALCULATE AVERAGES

import csv
from pathlib import Path

# Required dataset columns
USER_ID_COL = "User ID"
DEVICE_COL = "Device Model"
OS_COL = "Operating System"
BATTERY_COL = "Battery Drain (mAh/day)"
DATA_COL = "Data Usage (MB/day)"
BEHAVIOR_COL = "User Behavior Class"

# Required exported fields
EXPORT_FIELDS = [
    USER_ID_COL,
    DEVICE_COL,
    OS_COL,
    BATTERY_COL,
    DATA_COL
]

# Initialize counters and totals
total_records = 0
matching_os_count = 0
selected_count = 0

battery_total = 0.0
data_total = 0.0

battery_valid_count = 0
data_valid_count = 0

invalid_records = 0

# Output CSV
output_file = OUTPUT_FOLDER / "filtered_mobile_users.csv"

# Check dataset
if not INPUT_FILE.exists():
    raise FileNotFoundError(
        f"Dataset not found: {INPUT_FILE.resolve()}"
    )

# Read CSV line by line
with open(
    INPUT_FILE,
    "r",
    newline="",
    encoding="utf-8-sig"
) as file:

    reader = csv.DictReader(file)

    # Validate required columns
    if not reader.fieldnames:
        raise ValueError("Dataset has no column headers.")

    missing_columns = [
        col for col in [
            USER_ID_COL,
            DEVICE_COL,
            OS_COL,
            BATTERY_COL,
            DATA_COL,
            BEHAVIOR_COL
        ]
        if col not in reader.fieldnames
    ]

    if missing_columns:
        raise ValueError(
            f"Missing required columns: {missing_columns}"
        )

    # Create filtered CSV
    with open(
        output_file,
        "w",
        newline="",
        encoding="utf-8"
    ) as output:

        writer = csv.DictWriter(
            output,
            fieldnames=EXPORT_FIELDS
        )

        writer.writeheader()

        for row in reader:

            total_records += 1

            # Identify operating system
            operating_system = (
                row.get(OS_COL) or ""
            ).strip()

            # Select assigned operating system
            if operating_system.lower() != TARGET_OS.lower():
                continue

            matching_os_count += 1

            # Identify behavior class
            try:
                behavior_class = int(
                    float(row[BEHAVIOR_COL])
                )
            except (ValueError, TypeError, KeyError):
                invalid_records += 1
                continue

            # Select target behavior class
            if behavior_class != TARGET_BEHAVIOR_CLASS:
                continue

            selected_count += 1

            # Export specified fields
            writer.writerow({
                field: row.get(field, "")
                for field in EXPORT_FIELDS
            })

            # Calculate battery drain
            try:
                battery = float(row[BATTERY_COL])

                if battery >= 0:
                    battery_total += battery
                    battery_valid_count += 1

            except (ValueError, TypeError):
                invalid_records += 1

            # Calculate mobile data usage
            try:
                data = float(row[DATA_COL])

                if data >= 0:
                    data_total += data
                    data_valid_count += 1

            except (ValueError, TypeError):
                invalid_records += 1

# Calculate averages safely
average_battery_drain = (
    battery_total / battery_valid_count
    if battery_valid_count > 0
    else 0
)

average_data_usage = (
    data_total / data_valid_count
    if data_valid_count > 0
    else 0
)

# Display results
print(" MOBILE DEVICE USAGE PROFILING")

print("\nSTUDENT PARAMETERS")
print("SEED_DIGIT:", SEED_DIGIT)
print("ID_SUM:", ID_SUM)
print("Operating System:", TARGET_OS)
print("Behavior Class:", TARGET_BEHAVIOR_CLASS)

print("\nPROCESSING RESULTS")
print("Total records in dataset:", total_records)
print("Records matching OS:", matching_os_count)
print("Records in target behavior class:", selected_count)
print("Invalid data values:", invalid_records)

print("\nAVERAGE USAGE")
print(
    "Average battery drain:",
    round(average_battery_drain, 2),
    "mAh/day"
)

print(
    "Average data usage:",
    round(average_data_usage, 2),
    "MB/day"
)

print("\nEXPORT INFORMATION")
print("Exported CSV:", output_file)
print("Exported records:", selected_count)

print("\nCELL 3 COMPLETED SUCCESSFULLY!")
```

```python

# CELL 4: JSON SUMMARY AND CSV BACKUP

import json
import shutil

# Prepare summary information
summary = {
    "student_information": {
        "seed_digit": SEED_DIGIT,
        "id_sum": ID_SUM
    },

    "filter_criteria": {
        "operating_system": TARGET_OS,
        "user_behavior_class": TARGET_BEHAVIOR_CLASS
    },

    "dataset_information": {
        "total_records": total_records,
        "records_matching_os": matching_os_count,
        "records_in_target_class": selected_count,
        "invalid_data_values": invalid_records
    },

    "usage_summary": {
        "average_battery_drain_mAh_per_day": round(
            average_battery_drain, 2
        ),
        "average_data_usage_MB_per_day": round(
            average_data_usage, 2
        ),
        "valid_battery_records": battery_valid_count,
        "valid_data_records": data_valid_count
    },

    "export_information": {
        "exported_csv": output_file.name,
        "exported_record_count": selected_count
    }
}

# Create JSON summary
json_file = OUTPUT_FOLDER / "mobile_summary.json"

with open(
    json_file,
    "w",
    encoding="utf-8"
) as file:

    json.dump(summary, file, indent=4)

print("JSON SUMMARY")
print("JSON file created:", json_file)

# Verify exported CSV row count
with open(
    output_file,
    "r",
    newline="",
    encoding="utf-8-sig"
) as file:

    exported_reader = csv.DictReader(file)
    exported_count = sum(1 for row in exported_reader)

print("\nCOUNT VERIFICATION")
print("Selected record count:", selected_count)
print("Exported CSV row count:", exported_count)

if exported_count == selected_count:
    print("Count verification: SUCCESS")
else:
    print("Count verification: FAILED")

# Create CSV backup
backup_file = OUTPUT_FOLDER / "filtered_mobile_users_backup.csv"

shutil.copy2(output_file, backup_file)

# Verify backup
if backup_file.exists():

    original_size = output_file.stat().st_size
    backup_size = backup_file.stat().st_size

    print("\nBACKUP VERIFICATION")
    if original_size == backup_size:

        with open(output_file, "rb") as original:
            original_content = original.read()

        with open(backup_file, "rb") as backup:
            backup_content = backup.read()

        if original_content == backup_content:
            print("Backup verification: SUCCESS")
        else:
            print("Backup verification: FAILED")

    else:
        print("Backup verification: FAILED")

    print("Backup file:", backup_file)
    print("Backup file size:", backup_size, "bytes")

else:
    print("Backup file was not created.")

print("\nCELL 4 COMPLETED SUCCESSFULLY!")
```
